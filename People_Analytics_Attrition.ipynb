# 2. Clean and prepare

```python
df_clean = df.copy()
df_clean.info()
```

```python
df_clean = df_clean.set_index("EmployeeNumber")

constant_cols = ["EmployeeCount", "StandardHours", "Over18"]
df_clean = df_clean.drop(columns=constant_cols)

df_clean
```

```python
df_clean["Left"] = (df_clean["Attrition"] == "Yes").astype(int)
df_clean[["Attrition", "Left"]].head()
```

```python
df_clean["Left"].mean()
```

```python
satisfaction_map = {1: "Low", 2: "Medium", 3: "High", 4: "Very High"}
education_map = {1: "Below College", 2: "College", 3: "Bachelor", 4: "Master", 5: "Doctor"}
worklife_map = {1: "Bad", 2: "Good", 3: "Better", 4: "Best"}
performance_map = {1: "Low", 2: "Good", 3: "Excellent", 4: "Outstanding"}
df_clean["JobSatisfaction_label"] = df_clean["JobSatisfaction"].map(satisfaction_map)
df_clean["Education_label"] = df_clean["Education"].map(education_map)
df_clean["WorkLifeBalance_label"] = df_clean["WorkLifeBalance"].map(worklife_map)
df_clean["PerformanceRating_label"] = df_clean["PerformanceRating"].map(performance_map)
df_clean[["JobSatisfaction", "JobSatisfaction_label"]].head()
```

```python
sat_order = ["Low", "Medium", "High", "Very High"]
df_clean["JobSatisfaction_label"] = pd.Categorical(
    df_clean["JobSatisfaction_label"], categories=sat_order, ordered=True
)

df_clean["JobSatisfaction_label"].value_counts().sort_index()
```

```python
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[17, 25, 35, 45, 55, 60],
    labels=["18-25", "26-35", "36-45", "46-55", "56-60"]
)
df_clean["AgeGroup"].value_counts().sort_index()
```

```python
df_clean["TenureBand"] = pd.cut(
    df_clean["YearsAtCompany"],
    bins=[-1, 1, 3, 5, 10, 40],
    labels=["0-1 yrs", "2-3 yrs", "4-5 yrs", "6-10 yrs", "11+ yrs"]
)
df_clean["TenureBand"].value_counts().sort_index()
```

```python
df_clean["IncomeQuartile"] = pd.qcut(
    df_clean["MonthlyIncome"], q=4,
    labels=["Q1 (lowest)", "Q2", "Q3", "Q4 (highest)"]
)
df_clean["IncomeQuartile"].value_counts()
```

# 3. Descriptive analytics

```python
df_clean.groupby("Department")["Left"].mean()
```

```python
dept = df_clean.groupby("Department").agg(
    employees=("Left", "size"),
    leavers=("Left", "sum"),
    attrition_rate=("Left", "mean"),
)
dept["share_of_leavers"] = dept["leavers"] / dept["leavers"].sum()
dept
```

```python
company_rate = df_clean["Left"].mean()

def attrition_by(column):
    t = df_clean.groupby(column, observed=True).agg(
        employees=("Left", "size"),
        leavers=("Left", "sum"),
        attrition_rate=("Left", "mean"),
    )
    t["share_of_leavers"] = t["leavers"] / t["leavers"].sum()
    t["vs_company"] = t["attrition_rate"] / company_rate
    return t.sort_values("attrition_rate", ascending=False).round(3)

attrition_by("OverTime")
```

```python
attrition_by("JobRole")
attrition_by("AgeGroup")
attrition_by("TenureBand")
attrition_by("IncomeQuartile")
attrition_by("MaritalStatus")
attrition_by("BusinessTravel")
attrition_by("JobSatisfaction_label")
attrition_by("Gender")
```
